```python
def number_generator():
    for i in range(5):
        yield i

gen = number_generator()

for num in gen:
    print(num)
```

```python
def number_list():
    result = []
    for i in range(5):
        result.append(i)
    return result

nums = number_list()
for num in nums:
    print(num)
```

## Key Difference

**Generator (`yield`):**
- Values produced **one at a time** as needed
- Function pauses and resumes between yields
- Memory efficient - only one value in memory at a time
- Perfect for large datasets!

**List (`return`):**
- **All values** computed and stored immediately
- Function runs completely before returning
- Uses more memory - all values stored at once
- Fine for small datasets
